````python
from typing import Optional
def addTwoNumbers(l1: Optional[list] = None, l2: Optional[list] = None) -> Optional[list]:

    value = 0
    if l1 is not None:
        l1_word = ''
        i = -1
        while i > -(len(l1)+1):
            l1_word = f"{l1_word}{l1[i]}" 
            i -= 1

        value += int(l1_word)
    
    if l2 is not None:
        l2_word = ''
        i = -1
        while i > -(len(l2)+1):
            l2_word = f"{l2_word}{l2[i]}" 
            i -= 1
        value += int(l2_word)
    
    
    value_word = str(value)

    value_list = []
    i = -1
    while i > -(len(value_word) + 1):
        value_list.append(int(value_word[i]))
        i -= 1

    return(value_list)



# l1 = [2,4,3], l2= [5,6,4]
# 342 + 465 = 807

a = addTwoNumbers(l1=[0],l2=[0])
a
````

````python
from pyprojroot import here
import json
from dataclasses import dataclass
from analyst_agent.sft.pairs import build_pairs
from analyst_agent.agent import Output
import logging
import re

logger = logging.getLogger()

def extract_code(out: str) -> Output:
        """
        Extract executable Python code from a model response.

        The function removes any reasoning enclosed within </think>
        tags before attempting to extract code blocks. If no fenced
        code block is present but Python code appears to have been
        generated, the remaining text is treated as code.

        Args:
            out: Raw model output.

        Returns:
            Output containing both the raw response and extracted code.
        """

        # Remove reasoning section (if present)
        stage_1 = out.split('</think>')[-1].strip()

        # Only attempt extraction if executable code appears present
        if 'print(' in stage_1:

            # Look for fenced Python code blocks
            match = re.search(r"```[a-z]*\s*(.*?)```", stage_1, re.DOTALL)

            # If no fenced block exists, assume the remaining text is code
            if match is None:
                code = stage_1
            else:
                code = match.group(1).strip()

        else:
            logger.info('No code extracted')
            code = None

        return Output(
            raw=out,
            code=code
        )


@dataclass
class Pair:
    prompt: list[dict[str, str]]
    target: str

with open(here() / 'reports' / 'evaluation_2026-08-12_19-30-31.json', 'r') as f:
    raw_pairs = json.load(f)


cleaned_pairs = build_pairs(raw_pairs)

for pair in cleaned_pairs:
    output = extract_code(pair.target)
    assert output.code and output.code.strip(), "empty code"
    assert '<tool_call>' not in pair.target, "target would route to tool loop"
print(f"all {len(cleaned_pairs)} round-trip clean and route to run_task")



````
