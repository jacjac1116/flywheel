````python
from typing import Optional
def addTwoNumbers(l1: Optional[list] = None, l2: Optional[list] = None) -> Optional[list]:

    value = 0
    if l1 is not None:
        l1_word = ''
        i = -1
        while i > -(len(l1)+1):
            l1_word = f"{l1_word}{l1[i]}" 
            i -= 1

        value += int(l1_word)
    
    if l2 is not None:
        l2_word = ''
        i = -1
        while i > -(len(l2)+1):
            l2_word = f"{l2_word}{l2[i]}" 
            i -= 1
        value += int(l2_word)
    
    
    value_word = str(value)

    value_list = []
    i = -1
    while i > -(len(value_word) + 1):
        value_list.append(int(value_word[i]))
        i -= 1

    return(value_list)



# l1 = [2,4,3], l2= [5,6,4]
# 342 + 465 = 807

a = addTwoNumbers(l1=[0],l2=[0])
a
````
